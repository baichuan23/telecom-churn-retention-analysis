```python
from pathlib import Path
from importlib.metadata import version, PackageNotFoundError

project_root = Path(
    r"C:\Users\21129\Desktop\硕士申请材料\telecom-churn-retention-analysis"
)

packages = [
    "numpy",
    "pandas",
    "matplotlib",
    "seaborn",
    "scikit-learn",
    "joblib",
    "jupyter",
    "ipykernel",
]

requirements = []

for package in packages:
    try:
        requirements.append(f"{package}=={version(package)}")
    except PackageNotFoundError:
        print(f"Not installed: {package}")

output_path = project_root / "requirements.txt"
output_path.write_text("\n".join(requirements) + "\n", encoding="utf-8")

print(f"Saved: {output_path}")
print(output_path.read_text(encoding="utf-8"))
```

```python
ignore_path = project_root / ".gitignore"

rules = [
    ".ipynb_checkpoints/",
    "__pycache__/",
    "*.py[cod]",
    ".venv/",
    "venv/",
    ".env",
    ".env.*",
    "!.env.example",
    "Thumbs.db",
    ".DS_Store",
    "~$*",
]

existing = (
    ignore_path.read_text(encoding="utf-8")
    if ignore_path.exists()
    else ""
)

missing = [rule for rule in rules if rule not in existing.splitlines()]

if missing:
    separator = "\n" if existing and not existing.endswith("\n") else ""
    ignore_path.write_text(
        existing + separator + "\n".join(missing) + "\n",
        encoding="utf-8",
    )

print(f"Saved: {ignore_path}")
```

```python
folders = ["notebooks", "sql", "dashboard", "report"]

for folder in folders:
    folder_path = project_root / folder
    print(f"\n{folder}/")

    files = sorted(
        path for path in folder_path.rglob("*")
        if path.is_file()
        and ".ipynb_checkpoints" not in path.parts
    )

    if not files:
        print("  (empty)")
        continue

    for path in files:
        relative_path = path.relative_to(folder_path)
        size_kb = path.stat().st_size / 1024
        print(f"  {relative_path} ({size_kb:.1f} KB)")
```

```python
from pathlib import Path

print(f"Current working directory: {Path.cwd()}")

search_roots = [
    project_root,
    Path.cwd(),
    Path.home() / "Downloads",
    Path.home() / "Documents",
]

found = set()

for root in search_roots:
    if not root.exists():
        continue

    for pattern in ("*.ipynb", "*.pbix"):
        for path in root.rglob(pattern):
            if any(
                part in {".ipynb_checkpoints", ".venv", "venv"}
                for part in path.parts
            ):
                continue

            resolved = path.resolve()
            if resolved not in found:
                found.add(resolved)
                print(resolved)
```
